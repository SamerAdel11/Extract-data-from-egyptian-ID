```python
import pandas as pd
```

### 📁 Dataset Column Descriptions

- **`folder`**: Indicates the **original image name** (e.g., the full ID card image).
- **`filename`**: Represents the **cropped ID component** (e.g., `Id`, `First Name`, `Last Name`, `Address1`, `Address2`).
- **`text`**: Contains the **ground truth value** (i.e., the correct text content) for the corresponding cropped image.


```python
df=pd.read_excel('accuracy.xlsx')
df.head()
```

### Evaluation Criteria

- **`preprocessing_result`** and **`not_preprocessed_result`**:
  - A value of **`1`** indicates that the extracted data is **correct**.
  - A value of **`0`** indicates that the extracted data is **incorrect**.

- **`different_ocr_result`**:
  - A value of **`1`** means there is a **difference** between the OCR outputs with and without preprocessing.
  - A value of **`0`** means the outputs from both methods **match**.


```python
#############################################
### Extracted Data with image preprocessing
#############################################
```

```python
preprocessed_df=df.loc[df['preprocessing_result'] == 1]
```

```python
preprocessed_accuracy=preprocessed_df.groupby('filename').count().sort_values(by='folder',ascending=False)['folder']
preprocessed_accuracy
```

```python
#############################################
#### Extracted Data without image preprocessing
#############################################
```

```python
not_preprocessed_df=df.loc[df['not_preprocessed_result'] == 1]
```

```python
# Count the correct extracted values grouped by the Extracted Component
not_processed_accuracy=not_preprocessed_df.groupby('filename').count().sort_values(by='folder',ascending=False)['folder']
not_processed_accuracy
```

```python
# Convert the two accuracy series to one DataFrame
accuracy=pd.concat([preprocessed_accuracy,not_processed_accuracy],axis=1)
accuracy.columns=['preprocessed','not_preprocessed']
accuracy
```

### 🔍 Observation

The **non-preprocessed images** produced **better results** compared to the **preprocessed ones**.  
This suggests that the current preprocessing techniques tend to **degrade most images**, offering improvements in only a **few specific cases**.


```python
# Get the number of scanned images
images_number=len(df.loc[df['filename']=='Id.jpg'])
images_number
```

```python
#############################################
#### Calculate accuracy for not preprocessed images
#############################################
```

```python
for filename, row in accuracy.iterrows():
    name = filename.split('.')[0]
    acc = (row['not_preprocessed'] / images_number) * 100
    print(f"{name} accuracy is {acc.round(2)}%")
```

⚠️ **Note:**  
The overall low accuracies are mainly attributed to the **poor quality** of the image dataset.  

✅ **Highlight:**  
Despite this, the **ID number recognition** reached an accuracy of **71%**, which is a **notable and promising result**.
